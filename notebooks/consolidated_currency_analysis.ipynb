# Analysis for emergency liquidity for hard currency debt service across African sample

## Motivation and description of instrument

### Motivation

Currency mismatch between revenues from the sale of electricity in developing countries and the cost of hard currency debt service holds up many infrastructure finance transactions. When such hard currency credit has been provided for developing country electricity generation infrastructure, the currency risk has often been placed on the national utility - "offtaker" - through dollarised power purchase agreements (PPAs). This has led to an exacerbation of these utilities' financial woes from currency devaluations. This has had the knock-on effect of increasing offtaker risk for such projects and generating an additional impediment to the financing of electricity generation projects. Capital costs for such projects are thus affected two-fold by the currency mismatch, by the direct currency risk and by the induced offtaker risk. 

This proposal, analogous to the emergency liquidity facility of the EcoInvest Brazil platform, proposes to split the risk into an inflation risk component, to be shared by the final consumers and the utility (offtaker in the PPAs), and a real component, for which it is possible to structure affordable cover, as we shall see below.

### Simple model

The analysis below focuses on $H_{i,t} = \frac{P_{i,t}}{X_{i,t}}$ where, for country $i$ $P_i$ is the price level and $X_i$ is the nominal exchange rate, measured in USD/LCU (local currency units). As an index, this quantity measures the changes in the USD value of a fixed basket of goods in country $i$. If the price of electricity locally is made to grow at the overall national inflation rate, the index also measures the evolution of the USD value of a MWh of electricity.

Consider, for simplicity, a project that will generate a constant amount of energy $E$ from year $t_0$ to year $T$ at constant real unit costs $c$, excluding foreign currency debt service. If the price it receives per MWh is inflation adjusted, its revenues can be written as $R_t = P_{i,t}\cdot r \cdot E$ and its earnings before foreign currency debt service are $P_{i,t}\cdot(r - c) \cdot E$. That is the say, the nominal local currency cash flows available for debt service (CFADS) are equal to a constant growing at the rate of inflation and the CFADS in dollars are equal to $H_{i,t} \cdot (r-c) \cdot E$. 

For the sake of exposition, we will consider a stylised project in which all debt is in hard currency and $H_{i,t_0}\cdot (r-c) \cdot E= 100$ where $t_0$ is the first year of debt service. That is to say, at the initial exchange rate and price level 100 USD are available for foreign currency debt service. If $H$ decreases by 10\% only 90 USD are available for debt service. 

Usually project finance structures will require a debt service coverage ratio (DSCR) of 120-130\%, which in our stylised example would imply that debt service ought to be 76-83 USD each period. For ease of exposition we simplify further to a DSCR=1, assuming away any buffer to be used to cover forex swings. Notice that this also implies the project has zero expected profit, any decrease in $H$ creates a debt service shortfall (and technical default) and any increase in $H$ creates a surplus. This simplification is without loss of generality. The main result below on the affordability of an instrument structured to bear the real exchange rate risk is only strengthened if DSCR >1.

Back to our stylised project, any increase in $H_{i,t}$ over its value at $t_0$ amounts to a profit flow equal, in undiscounted dollar terms, to the growth rate $g_{i,t}=\frac{H_{i,t}}{H_{i,t_0}}-1$ times 100. Any decrease relative to the $t_0$ value equals the analogous loss.

Given a discount rate $\rho$, the integral
$$
    \pi_{i,t_0-T} =  \int_{t_0}^T \frac{g_{i,t}}{(1+\rho)^t}dt
$$

is a measure of the present value of the lifetime loss (or profit) experienced by such a project operating between $t_0$ and $T$, due to the variability in the price level and the exchange rate. The actual profit/loss is $100\cdot \pi_{i,t_0-T}$, on account of the debt service cost being constant and equal to 100. If the debt service profile were not constant, the value would have to remain in the integral. Notice that any repayment profile that is more back-loaded (as is common in such transactions) will put more weight on the more highly discounted later periods and thus decrease the value of $pi$ in absolute terms. 

### Instrument

Now consider the following emergency liquidity insurance product provided by the financial institution $\Gamma$. Whenever $g_{i,t}$ is negative, $\Gamma$ provides the project with a loan worth $100 \cdot g_{i,t}$ USD, so that the project can meet its debt service obligations. In the alternate cases, when $g_{i,t}$ is positive, the project puts the surplus is has into an interest bearing dollar escrow account to be used to pay back $\Gamma$ at $T$. The result is that, whenever $\pi_{i,t_0-T}>0$, $\Gamma$ is repaid with interest, and whenever $\pi_{i,t_0-T}<0$, $\Gamma$ sits on a loss worth $100\cdot \pi_{i,t_0-T}$ in present value.

We can think of access to such emergency liquidity as an insurance product, for which $\Gamma$ charges a premium to cover its expected loss. Abstracting from solvency capital requirements and profits, the annual premium is simply the annuity that equals the expected loss. Defining $\alpha_{\rho,t}$ the present value of a 1 USD annuity for $t$ years at a discount rate of $\rho$ we get $\alpha_{\rho,t} = (1-(1-\rho)^{-t})/\rho$. Multiplying by $\alpha_{\rho,t}$ converts an annuity into the net present value, and dividing by $\alpha_{\rho,t}$ converts a present value into an annuity. 

The actuarially fair annual premium for this insurance product is thus $100\frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha_{\rho,T-t_0}}$. Recall that the $100$ stems from our simplifying assumption of annual debt service payments of 100 USD, so one can simply equate $\frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha_{\rho,T-t_0}}$ to the annual premium as a proportion of the annual debt service payments. Since the debt service payments in our simple example are constant, they also equal the annuity equivalent to the borrowed debt stock (take a moment to reflect on why this is so). So if we want to convert the annual premium into a percentage of the total USD borrowing, we simply note that, because debt service is $T-t_0$ installations of 100 USD, the amount borrowed must be $D = 100 \cdot \alpha_{\rho,T-t_0}$. The actuarially fair premium is thus
$$
  \Psi =  D \cdot \frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha^2_{\rho,T-t_0}}
$$
Put differently, $\Delta r = \frac{\Psi}{D} = \frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha^2_{\rho,T-t_0}}$ is equal to the additional interest rate, measured as a proportion of the principal, needed to cover the insurance premium.

## The terminal liability

The function `compute_terminal_residuals` in `data_analysis_functions.py` returns a dataframe in which each row calculates the profit/liability $\pi_{i,t_0-T}$ for a given time window and country in the data. In the call below it takes every 15 year window for which there is data for a country, and calculates $\pi_{i,t_0-T}$. As you can see, for Angola the dataset starts in 2006 and ends in 2023, so there are only 4 (overlapping) windows, 2006-2020, 2007-2021, 2008-2022, and 2009-2023. Countries with more data will have more windows. 

For each of these windows, of which there are 452 in 52 countries for which there is data, `compute_terminal_residuals` computes $\pi_{i,t_0-T}$ (in line 174) as well as the gain/loss as a fraction of the debt stock $\frac{\pi_{i,t_0-T}}{\alpha^2_{\rho,T-t_0}}$ (in line 176), and the "loss", which is simply the minumum between the latter and zero. (The dataframe also returns the value of $H$ at the beginning of the window). 

The calculation below shows that the annual premium for a diversified portfolio over the entire African continent would result in an additional 0.19\% interest rate, driven by the fact that only in 16\% of cases a liability is incurred by $\Gamma$, in all other cases they are made whole.

```python
residuals = compute_terminal_residuals(hybrid, years=15, discount_rate=0.08, window_mode="rolling")
print(residuals.head())

expected_loss = residuals["loss"].mean()
prob_positive = (residuals["loss"] < 0).mean()

print(f"Annual premium: {expected_loss*100:.2f}% of debt stock")
print(f"Probability of Loss: {prob_positive*100:.0f}%")
```

## Country variation

The pooled premium above averages over all rolling 15-year windows across countries in the sample. Each country contributes up to 20 windows (start years 1990–2009), fewer for countries whose data begins later. This section examines how the premium varies across countries and exchange-rate regimes.

```python
# Country-level statistics
residuals['regime'] = residuals['country'].map(fx_regime).fillna('Flexible')

country_stats = residuals.groupby('country').agg(
    n_windows=('loss', 'count'),
    premium=('loss', 'mean'),
    prob_loss=('loss', lambda x: (x < 0).mean()),
    e_loss_given_loss=('loss', lambda x: x[x < 0].mean() if (x < 0).any() else 0.0),
    worst=('loss', 'min'),
    best=('net_pv_by_debt_stock', 'max'),
).reset_index()
country_stats['regime'] = country_stats['country'].map(fx_regime).fillna('Flexible')
country_stats = country_stats.sort_values('premium')

pooled_premium = residuals['loss'].mean()

display_df = country_stats[['country', 'regime', 'n_windows', 'premium', 'prob_loss', 'e_loss_given_loss', 'worst', 'best']].copy()
display_df.columns = ['Country', 'Regime', 'Windows', 'Premium (%)', 'P(loss)', 'E[loss|loss] (%)', 'Worst (%)', 'Best (%)']
for col in ['Premium (%)', 'E[loss|loss] (%)', 'Worst (%)', 'Best (%)']:
    display_df[col] = display_df[col] * 100
display_df['P(loss)'] = display_df['P(loss)'] * 100

print(f"Pooled annual premium: {pooled_premium*100:.3f}% of debt stock")
print(f"Countries: {len(country_stats)}, Total windows: {residuals.shape[0]}\n")
print(display_df.to_string(index=False, float_format='%.3f'))
```

## Correlation and diversification

The benefit of a pan-African pooling facility depends on how correlated real exchange rate movements are across countries. High within-group correlation (especially for EUR-linked countries sharing a common anchor) limits diversification, while low cross-group correlation enhances it.

```python
# Correlation heatmap of annual H changes, ordered by regime
wide = hybrid.pivot(index='year', columns='country', values='fx_index')
returns = wide.pct_change().dropna()
corr = returns.corr()

# Order countries by regime
eur_countries = sorted([c for c in corr.columns if fx_regime.get(c) == 'EUR-linked'])
zar_countries = sorted([c for c in corr.columns if fx_regime.get(c) == 'ZAR-linked'])
flex_countries = sorted([c for c in corr.columns if c not in eur_countries + zar_countries])
ordered = eur_countries + zar_countries + flex_countries

# Melt to long format
corr_ordered = corr.loc[ordered, ordered]
corr_long = corr_ordered.reset_index().melt(id_vars='country', var_name='country2', value_name='correlation')
corr_long = corr_long.rename(columns={'country': 'country1'})

heatmap = alt.Chart(corr_long).mark_rect().encode(
    x=alt.X('country2:N', sort=ordered, title=None),
    y=alt.Y('country1:N', sort=ordered, title=None),
    color=alt.Color('correlation:Q',
                    scale=alt.Scale(scheme='redblue', domain=[-0.5, 1], domainMid=0),
                    title='Corr'),
    tooltip=['country1', 'country2', alt.Tooltip('correlation:Q', format='.2f')]
).properties(width=500, height=500, title='Pairwise Correlation of Annual Real Exchange Rate Changes')

# Regime boundary lines
n_eur = len(eur_countries)
n_zar = len(zar_countries)

boundary_data = pd.DataFrame({'pos': [n_eur - 0.5, n_eur + n_zar - 0.5]})
vrules = alt.Chart(boundary_data).mark_rule(color='black', strokeWidth=1.5).encode(x='pos:Q')
hrules = alt.Chart(boundary_data).mark_rule(color='black', strokeWidth=1.5).encode(y='pos:Q')

(heatmap + vrules + hrules).configure_axis(labelFontSize=6)
```

```python
# Mean pairwise correlation statistics
def mean_pairwise_corr(corr_matrix, group):
    sub = corr_matrix.loc[group, group]
    mask = np.triu(np.ones(sub.shape, dtype=bool), k=1)
    vals = sub.where(mask).stack()
    return vals.mean(), vals.std(), len(vals)

def mean_cross_corr(corr_matrix, g1, g2):
    vals = corr_matrix.loc[g1, g2].stack()
    return vals.mean(), vals.std(), len(vals)

print("Mean pairwise correlation of annual H changes")
print("=" * 60)

corr_rows = []
m, s, n = mean_pairwise_corr(corr, eur_countries)
corr_rows.append({'Group': 'Within EUR-linked', 'Mean corr': m, 'Std': s, 'Pairs': n})

if len(zar_countries) > 1:
    m, s, n = mean_pairwise_corr(corr, zar_countries)
    corr_rows.append({'Group': 'Within ZAR-linked', 'Mean corr': m, 'Std': s, 'Pairs': n})

m, s, n = mean_pairwise_corr(corr, flex_countries)
corr_rows.append({'Group': 'Within Flexible', 'Mean corr': m, 'Std': s, 'Pairs': n})

m, s, n = mean_cross_corr(corr, eur_countries, flex_countries)
corr_rows.append({'Group': 'EUR-linked vs Flexible', 'Mean corr': m, 'Std': s, 'Pairs': n})

if len(zar_countries) > 0:
    m, s, n = mean_cross_corr(corr, zar_countries, flex_countries)
    corr_rows.append({'Group': 'ZAR-linked vs Flexible', 'Mean corr': m, 'Std': s, 'Pairs': n})
    m, s, n = mean_cross_corr(corr, eur_countries, zar_countries)
    corr_rows.append({'Group': 'EUR-linked vs ZAR-linked', 'Mean corr': m, 'Std': s, 'Pairs': n})

corr_table = pd.DataFrame(corr_rows)
print(corr_table.to_string(index=False, float_format='%.3f'))
```

```python
# Diversification benefit: pooled vs average country-level
avg_country_premium = country_stats['premium'].mean()
avg_country_ploss = country_stats['prob_loss'].mean()

pooled_ploss = (residuals['loss'] < 0).mean()

print("Diversification benefit")
print("=" * 60)
print(f"{'Metric':<35} {'Pooled':>12} {'Avg country':>12}")
print("-" * 60)
print(f"{'Annual premium (% debt stock)':<35} {pooled_premium*100:>12.3f} {avg_country_premium*100:>12.3f}")
print(f"{'P(loss)':<35} {pooled_ploss*100:>12.1f}% {avg_country_ploss*100:>12.1f}%")
print()
print("The pooled premium equals the average country premium (the expectation")
print("is additive). The diversification benefit comes from variance reduction:")
print("a facility covering many countries faces a tighter loss distribution,")
print("reducing capital requirements.")
```

```python
# Bootstrap confidence intervals (country-block resampling, 10,000 draws)
def bootstrap_premium(res_df, n_boot=10000, seed=42):
    rng = np.random.default_rng(seed)
    countries = res_df['country'].unique()
    n = len(countries)
    country_losses = {c: res_df.loc[res_df['country'] == c, 'loss'].values for c in countries}
    boot = np.empty(n_boot)
    for b in range(n_boot):
        sampled = rng.choice(countries, size=n, replace=True)
        all_losses = np.concatenate([country_losses[c] for c in sampled])
        boot[b] = all_losses.mean()
    point = res_df['loss'].mean()
    return {'point': point, 'se': boot.std(),
            'ci_lo': np.percentile(boot, 2.5), 'ci_hi': np.percentile(boot, 97.5),
            'distribution': boot}

# Run bootstrap for overall and per-regime
boot_results = {}
boot_results['All'] = bootstrap_premium(residuals)
for regime in ['EUR-linked', 'ZAR-linked', 'Flexible']:
    boot_results[regime] = bootstrap_premium(residuals[residuals['regime'] == regime])

# Print summary
print("Bootstrap confidence intervals (10,000 country-block resamples)")
print("=" * 70)
for group, b in boot_results.items():
    print(f"{group:>12}: premium = {b['point']*100:.3f}%  "
          f"SE = {b['se']*100:.3f}%  "
          f"95% CI = [{b['ci_lo']*100:.3f}%, {b['ci_hi']*100:.3f}%]")

# Pre-bin bootstrap distributions in pandas to avoid Altair's 5000-row limit
group_order = ['All', 'EUR-linked', 'ZAR-linked', 'Flexible']
group_colors = ['steelblue', '#4c78a8', '#f58518', '#54a24b']

binned_rows = []
annot_rows = []
for group, b in boot_results.items():
    dist = b['distribution'] * 100
    counts, edges = np.histogram(dist, bins=50)
    centers = (edges[:-1] + edges[1:]) / 2
    for c, cnt in zip(centers, counts):
        if cnt > 0:
            binned_rows.append({'group': group, 'bin_center': c, 'count': int(cnt)})
    annot_rows.append({'group': group, 'x': b['point'] * 100, 'type': 'point'})
    annot_rows.append({'group': group, 'x': b['ci_lo'] * 100, 'type': 'ci'})
    annot_rows.append({'group': group, 'x': b['ci_hi'] * 100, 'type': 'ci'})

binned_df = pd.DataFrame(binned_rows)
annot_df = pd.DataFrame(annot_rows)

# Merge annotations into binned data for faceted layering
binned_df = binned_df.merge(
    annot_df[annot_df['type'] == 'point'][['group', 'x']].rename(columns={'x': 'point_est'}),
    on='group'
)
binned_df = binned_df.merge(
    annot_df[annot_df['type'] == 'ci'].groupby('group')['x'].agg(['min', 'max']).rename(
        columns={'min': 'ci_lo', 'max': 'ci_hi'}),
    on='group'
)

base = alt.Chart(binned_df)

hist_chart = base.mark_bar(opacity=0.8).encode(
    x=alt.X('bin_center:Q', title='Annual premium (% of debt stock)'),
    y=alt.Y('count:Q', title='Frequency'),
    color=alt.Color('group:N', scale=alt.Scale(domain=group_order, range=group_colors), legend=None)
)

point_rules = base.mark_rule(color='red', strokeDash=[4, 4], strokeWidth=1.5).encode(
    x='point_est:Q'
)

ci_lo_rules = base.mark_rule(color='gray', strokeDash=[2, 2], strokeWidth=1).encode(
    x='ci_lo:Q'
)

ci_hi_rules = base.mark_rule(color='gray', strokeDash=[2, 2], strokeWidth=1).encode(
    x='ci_hi:Q'
)

(hist_chart + point_rules + ci_lo_rules + ci_hi_rules).facet(
    facet=alt.Facet('group:N', title='Group', sort=group_order),
    columns=2
).properties(title='Bootstrap Distributions of Annual FX Risk Premium')
```

### Visualizing the residuals for a single country

To build intuition for what `compute_terminal_residuals` is summing, it helps to look at one country window by window. Below we take Morocco and the first 15-year window beginning in **1995**. We draw a dashed reference line at the value of $H$ at the window's start and shade the gap between $H$ and that line: **green where $H$ is above the start level** (a gain for $\Gamma$) and **red where it is below** (a loss). The terminal residual is the discounted net of these two areas, $\pi/\alpha^2$ — so when green outweighs red the residual is positive, and when red dominates it is negative.

The bottom panel plots **one residual bar per annual window**, each positioned at the year the window *matures*, i.e. `window_start + 15`: the 1995 window's bar sits at 2010, the 1996 window's at 2011, and so on through the last full window. Because the windows are rolling and overlap heavily, neighbouring bars move smoothly; the first one (the shaded 1995–2009 window) is slightly negative because the deep 1995–2004 dip just outweighs the 2006–2009 recovery.

```python
# --- Morocco: first 15-yr window (1995) shaded, plus a residual bar for every window ---
COUNTRY, YEARS, DISCOUNT, STEP, FIRST_START = "MAR", 15, 0.08, 1, 1995

g = hybrid[hybrid.country == COUNTRY].sort_values("year").reset_index(drop=True)
af = annuity_factor(YEARS, DISCOUNT)


def window_residual(g, start_year, years=YEARS, r=DISCOUNT):
    """pi / alpha**2 for the `years`-long window beginning in `start_year`."""
    win = g[(g.year >= start_year) & (g.year < start_year + years)]
    if len(win) < years:
        return None
    base = float(win.fx_index.iloc[0])
    git = (win.fx_index.to_numpy() / base) - 1.0          # deviation from start level
    t = np.arange(1, years + 1, dtype=float)
    return float(np.sum(git * (1.0 + r) ** (-t))) / af ** 2


# residual for every annual window start, plotted where the window matures (start + 15)
starts = range(FIRST_START, int(g.year.max()) - YEARS + 2, STEP)
bars = pd.DataFrame([{"window_start": s, "matures": s + YEARS,
                      "residual": window_residual(g, s)} for s in starts]).dropna()
bars["sign"] = np.where(bars.residual >= 0, "gain", "loss")

# green = area above the start level, red = area below it, for the FIRST window only.
# Interpolate H on a fine grid so the colour split is clean where H crosses the level.
fw = g[(g.year >= FIRST_START) & (g.year < FIRST_START + YEARS)]
ref = float(fw.fx_index.iloc[0])
xs = np.linspace(fw.year.min(), fw.year.max(), 400)
hs = np.interp(xs, fw.year, fw.fx_index)
shade = pd.DataFrame({"year": xs, "ref": ref,
                      "above": np.maximum(hs, ref), "below": np.minimum(hs, ref)})

xscale = alt.Scale(domain=[g.year.min(), g.year.max() + 1])

line = alt.Chart(g).mark_line(color="#222", strokeWidth=2).encode(
    x=alt.X("year:Q", scale=xscale, axis=alt.Axis(format="d", title=None)),
    y=alt.Y("fx_index:Q", title="hybrid index H"))
green = alt.Chart(shade).mark_area(opacity=0.4, color="#2ca02c").encode(
    x="year:Q", y="ref:Q", y2="above:Q")
red = alt.Chart(shade).mark_area(opacity=0.4, color="#d62728").encode(
    x="year:Q", y="below:Q", y2="ref:Q")
refrule = alt.Chart(shade).mark_line(strokeDash=[4, 4], color="#444").encode(
    x="year:Q", y="ref:Q")
top = (green + red + refrule + line).properties(
    width=660, height=300,
    title=f"{COUNTRY}: H with {FIRST_START}–{FIRST_START + YEARS - 1} window shaded "
          "(green above start = gain, red below = loss)")

# one bar per annual window; values are in the tooltip (labels would overlap at this density)
zero = alt.Chart(pd.DataFrame({"y": [0]})).mark_rule(color="#888").encode(y="y:Q")
bar = alt.Chart(bars).mark_bar(width=10).encode(
    x=alt.X("matures:Q", scale=xscale,
            axis=alt.Axis(format="d", title="year window matures (start + 15)")),
    y=alt.Y("residual:Q", title="terminal residual"),
    color=alt.Color("sign:N", scale=alt.Scale(domain=["gain", "loss"],
                    range=["#2ca02c", "#d62728"]), legend=None),
    tooltip=["window_start", "matures", alt.Tooltip("residual:Q", format="+.4f")])
bottom = (zero + bar).properties(width=660, height=170)

alt.vconcat(top, bottom).resolve_scale(x="shared")
```

### A well-designed single-country visualization

The cell above is faithful to the request but hardwired to Morocco and to one shaded window. The version below generalizes it into a small reusable function, `plot_country_residuals(country, ...)`, and fixes the two weaknesses of the first draft:

1. **Honest axes instead of a magic offset.** The residuals (order $10^{-2}$) and the index $H$ (order $10^{2}$) live on incompatible scales, so rather than scaling the bars by an arbitrary factor and hanging them under the line, they get their own panel with a real, labelled axis. The two panels share the year axis (`resolve_scale(x="shared")`), so a bar still sits directly under the end of the window it summarizes.

2. **One window at a time, by choice.** With an annual bar for every window, shading all of them at once would overplot. Instead the bottom panel is the index into the windows: **click any bar** and the top panel shades that window (green above its start level, red below), draws the dashed start-level reference across the window span, and the selected bar is shown at full opacity while the others dim. This ties each residual directly to the green-minus-red area that produced it.

The function exposes `years`, `discount_rate`, `step` and `first_start` so the same view works for any country and any windowing scheme. It defaults to `step=1` (a bar for every annual rolling window); pass e.g. `step=5` to thin the bars out.

```python
def plot_country_residuals(country, years=15, discount_rate=0.08, step=1,
                           first_start=None, df=hybrid, grid=300, ypad_frac=0.05):
    """Single-country view of the 15-yr terminal residuals.

    Top panel:  the hybrid index H, re-based to 100 in the country's first year.
                The y-axis does not start at zero -- it is padded just enough to
                keep the whole series in view.
    Bottom panel: one bar per window (annual by default), plotted at the year the
                  window ends (window_start + `years` - 1), green if the residual is
                  positive, red if negative.
    Interaction: click a bar to shade its window in the top panel -- green where
                 H sits above the window's start level (a gain), red where below
                 (a loss). The residual is the discounted net of those two areas
                 expressed per unit of debt stock, i.e. pi / alpha**2.
    """
    g = df[df.country == country].sort_values("year").reset_index(drop=True)
    # re-base to an index = 100 in the first displayed year (residuals are ratio-based,
    # so this rescaling leaves the bars unchanged and keeps the shading consistent)
    g["fx_index"] = g["fx_index"] / g["fx_index"].iloc[0] * 100.0
    af = annuity_factor(years, discount_rate)
    y0, y1 = int(g.year.min()), int(g.year.max())
    if first_start is None:
        first_start = y0
    starts = range(first_start, y1 - years + 2, step)

    bar_rows, shade_rows = [], []
    for s in starts:
        win = g[(g.year >= s) & (g.year < s + years)]
        if len(win) < years:
            continue
        ref = float(win.fx_index.iloc[0])
        git = (win.fx_index.to_numpy() / ref) - 1.0
        t = np.arange(1, years + 1, dtype=float)
        resid = float(np.sum(git * (1.0 + discount_rate) ** (-t))) / af ** 2
        # window covers years s .. s + years - 1; the bar sits at the window's end year
        bar_rows.append({"window_start": s, "window_end": s + years - 1, "residual": resid,
                         "sign": "gain" if resid >= 0 else "loss",
                         "label": f"{s}–{s + years - 1}"})
        # densify H within the window so the green/red split is clean at crossings
        xs = np.linspace(win.year.min(), win.year.max(), grid)
        hs = np.interp(xs, win.year, win.fx_index)
        for x, h in zip(xs, hs):
            shade_rows.append({"window_start": s, "year": x, "ref": ref,
                               "above": max(h, ref), "below": min(h, ref)})

    bars = pd.DataFrame(bar_rows)
    shade = pd.DataFrame(shade_rows)
    xscale = alt.Scale(domain=[y0, y1 + 1])
    # y-axis: don't start at zero; pad just enough to keep the full series in view
    ypad = ypad_frac * (g.fx_index.max() - g.fx_index.min())
    yscale = alt.Scale(domain=[g.fx_index.min() - ypad, g.fx_index.max() + ypad],
                       zero=False, nice=False)
    sel = alt.selection_point(fields=["window_start"], on="click", empty=False,
                              value=[{"window_start": bars.window_start.iloc[0]}])

    line = alt.Chart(g).mark_line(color="#222", strokeWidth=2).encode(
        x=alt.X("year:Q", scale=xscale, axis=alt.Axis(format="d", title=None)),
        y=alt.Y("fx_index:Q", title="hybrid index H (=100 in first year)", scale=yscale))
    green = alt.Chart(shade).transform_filter(sel).mark_area(
        opacity=0.4, color="#2ca02c").encode(x="year:Q", y=alt.Y("ref:Q", scale=yscale), y2="above:Q")
    red = alt.Chart(shade).transform_filter(sel).mark_area(
        opacity=0.4, color="#d62728").encode(x="year:Q", y=alt.Y("below:Q", scale=yscale), y2="ref:Q")
    refrule = alt.Chart(shade).transform_filter(sel).mark_line(
        strokeDash=[4, 4], color="#444").encode(x="year:Q", y=alt.Y("ref:Q", scale=yscale))
    top = (green + red + refrule + line).properties(
        width=660, height=300,
        title=alt.TitleParams(
            f"{country}: hybrid index H with selectable {years}-yr window",
            subtitle="green = H above window start (gain) | red = below (loss); "
                     "click a bar to change window"))

    zero = alt.Chart(pd.DataFrame({"y": [0]})).mark_rule(color="#888").encode(y="y:Q")
    bar = alt.Chart(bars).mark_bar(width=10).encode(
        x=alt.X("window_end:Q", scale=xscale,
                axis=alt.Axis(format="d", title=f"year window ends (start + {years - 1})")),
        y=alt.Y("residual:Q", title="terminal residual  (pi / alpha²)"),
        color=alt.Color("sign:N", scale=alt.Scale(domain=["gain", "loss"],
                        range=["#2ca02c", "#d62728"]), legend=None),
        opacity=alt.condition(sel, alt.value(1.0), alt.value(0.35)),
        tooltip=[alt.Tooltip("label:N", title="window"),
                 alt.Tooltip("residual:Q", format="+.4f")]).add_params(sel)
    bottom = (zero + bar).properties(width=660, height=170)

    return alt.vconcat(top, bottom).resolve_scale(x="shared").configure_view(strokeWidth=0)


plot_country_residuals("MAR", first_start=1995, step=1)
```

## Event based analysis

The analysis also covers a more event based approach that identifies event by the size of the peak to trough devaluation, and bookends them when $H$ returns to within a certain range of the initial value. This can be useful to explain and explore the dataset and will be described further below.
