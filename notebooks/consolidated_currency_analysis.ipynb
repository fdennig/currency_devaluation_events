# Analysis for emergency liquidity for hard currency debt service across African sample

## Motivation and description of instrument

Currency mismatch between revenues from the sale of electricity in developing countries and the debt service for infrastructure financed in hard currency causes fewer such transactions to take place, and when they do take place, the currency risk has often been placed on the national utility "offtaker" in the power purchase agreements. This has led to an exacerbation of these utilities financial woes by currency devaluations, thus increasing offtaker risk for such projects and generating an additional impediment to the financing of electricity generation projects. Capital costs for such projects are thus affected two-fold by the currency mismatch, by the direct currency risk and by the induced offtaker risk. 

This proposal, analogous to the emergency liquidity facility of the EcoInvest Brazil platform, proposes to split the risk into an inflation risk component, to be shared by the final consumers and the utility (offtaker in the PPAs), and a real component, for which it is possible to structure affordable cover. 

The analysis below focuses on $H_{i,t} = \frac{P_{i,t}}{X_{i,t}}$ where, for country $i$ $P_i$ is the price level and $X_i$ is the nominal exchange rate, measured in USD/LCU (local currency units). As an index, this quantity measures the changes in the USD value of a fixed basket of goods in country $i$. If the price of electricity locally is made to grow at the overall national inflation rate, the index measures the evolution of the USD value from the sale of a MWh of electricity.

Consider, for simplicity a project that will generate a constant amount of energy $E_t$ for $T$ years at constant real costs $c$, excluding foreign currency debt service. If the price it receives per MWh is inflation adjusted, its revenues can be written as $R_t = P_{i,t}\cdot r$ and its earnings before foreign currency debt service are $P_{i,t}\cdot(r - c)$. That is the say, the nominal local currency cash flows available for debt service (CFADS) are equal to a constant growing at the rate of inflation and the CFADS in dollars are equal to $H_{i,t} \cdot (r-c)$. 

For the sake of exposition, we will consider a stylised project in which all debt is in hard currency and $H_{i,t_0}\cdot (r-c) = 100$ where $t_0$ is the first of $T$ repayment periods. That is to say, at the initial exchange rate and price level 100 USD are available for foreign currency debt service. If $H$ decreases by 10\% only 90 USD are available for debt service. 

Usually project finance structures will require a debt service coverage ratio (DBCR) of 120-130\%, which in our stylised example would imply that debt service is 76-83 USD each period. Here again we simplify to a DBCR=1, assuming away any buffer to be used to cover forex swings. Notice that this also implies the project has zero expected profit and any decrease in $H$ creates a debt service shortfall (and technical default) and any increase in $H$ creates a surplus. This is without loss of generality, since the logic of the proposed instrument translates seemlessly to DBCR >1.

Back with our stylised project, any increase in $H_{i,t}$ over its value at $t_0$ amounts to a profit flow equal, in undiscounted dollar terms, to the growth rate $g_{i,t}=\frac{H_{i,t}}{H_{i,t_0}}-1$ times 100. Any decrease relative to the $t_0$ value equals the analogous loss.

Given a discount rate $\rho$, the integral
$$
    \pi_{i,t_0-T} =  \int_{t_0}^T \frac{g_{i,t}}{(1+\rho)^t}dt
$$

is equal to the present value of the lifetime loss (or profit) divided by 100, experienced by such a project operating between $t_0$ and $T$, due to the variability in the price level and the exchange rate.

Now consider the following emergency liquidity insurance product provided by the financial institution $\Gamma$. Whenever $g_{i,t}$ is negative, $\Gamma$ provides the project with a loan worth $100 \cdot g_{i,t}$ USD, so that the project can meet its debt service obligations. In the alternate cases, when $g_{i,t}$ is positive, the project puts the surplus is has into an interest bearing dollar escrow account to be used to pay back $\Gamma$ at $T$. The result is that, whenever $\pi_{i,t_0-T}>0$, $\Gamma$ is repaid with interest, and whenever $\pi_{i,t_0-T}<0$, $\Gamma$ sits on a loss worth $100\cdot \pi_{i,t_0-T}$ in present value.

We can think of access to such emergency liquidity as an insurance product, for which $\Gamma$ charges a premium to cover its expected loss. Abstracting from solvency capital requirements and profits, the annual premium is simply the annuity that equals the expected loss. Defining $\alpha_{\rho,t}$ the present value of a 1 USD annuity for $t$ years at a discount rate of $\rho$ we get $\alpha_{\rho,t} = (1-(1-\rho)^{-t})/\rho$. Multiplying by $\alpha_{\rho,t}$ converts an annuity into the net present value, and dividing by $\alpha_{\rho,t}$ converts a present value into an annuity. 

The actuarially fair annual premium for this insurance product is thus $100\frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha_{\rho,T-t_0}}$. Recall that the $100$ stems from our simplifying assumption of annual debt service payments of 100 USD, so one can simply equate $\frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha_{\rho,T-t_0}}$ to the annual premium as a proportion of the annual debt service payments. Since the debt service payments in our simple example are constant, they also equal the annuity equivalent to the borrowed debt stock (take a moment to reflect on why this is so). So if we want to convert the annual premium into a percentage of the total USD borrowing, we simply note that, because debt service is $T-t_0$ installations of 100 USD, the amount borrowed must be $D = 100 \cdot \alpha_{\rho,T-t_0}$. The actuarially fair premium is thus
$$
  \Psi =  D \cdot \frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha^2_{\rho,T-t_0}}
$$
Put differently, $\Delta r = \frac{\Psi}{D} \cdot 100 = \frac{\mathbb{E}[\pi_{i,t_0-T}]}{\alpha^2_{\rho,T-t_0}} \cdot 100$ is equal to the additional interest rate, measured in percent, needed to cover the insurance premium.

## The terminal liability

The function `compute_terminal_residuals` in `data_analysis_functions.py` returns a dataframe in which each row calculates the profit/liability $\pi_{i,t_0-T}$ for a given time window and country in the data. In the call below it takes every 15 year window for which there is data for a country, and calculates $\pi_{i,t_0-T}$. As you can see, for Angola the dataset starts in 2006 and ends in 2023, so there are only 4 (overlapping) windows, 2006-2020, 2007-2021, 2008-2022, and 2009-2023. Countries with more data will have more windows. 

For each of these windows, of which there are 452 in 52 countries for which there is data, `compute_terminal_residuals` computes $\pi_{i,t_0-T}$ (in line 174) as well as the gain/loss as a fraction of the debt stock $\frac{\pi_{i,t_0-T}}{\alpha^2_{\rho,T-t_0}}$ (in line 176), and the "loss", which is simply the minumum between the latter and zero. (The dataframe also returns the value of $H$ at the beginning of the window). 

The calculation below shows that the annual premium for a diversified portfolio over the entire African continent would result in an additional 0.19\% interest rate, driven by the fact that only in 16\% of cases a liability is incurred by $\Gamma$, in all other cases they are made whole.

```python
residuals = compute_terminal_residuals(hybrid, years=15, discount_rate=0.1, window_mode="rolling")
print(residuals.head())

expected_loss = residuals["loss"].mean()
prob_positive = (residuals["loss"] < 0).mean()

print(f"Annual premium: {expected_loss*100:.2f}% of debt stock")
print(f"Probability of Loss: {prob_positive*100:.0f}%")
```
